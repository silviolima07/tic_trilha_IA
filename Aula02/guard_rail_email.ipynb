### No site da OpenRouter, na api key criada, foi criado uma regra de guard-rail.
### Objetivo: para esconder endereços de email e colocar no lugar um placeholder.
###
-  "content": "Título: Inteligência Artificial na Saúde  \nAutor: João da Silva  \nContato: [EMAIL]"

```python
import os
import json
import requests
from dotenv import load_dotenv

load_dotenv(override=True)

openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
```

```python
texto_teste = """
Título: Inteligência Artificial na Saúde
Autor: João da Silva
Contato: joao.silva@universidade.edu.br
"""

resposta = requests.post(
    url="https://openrouter.ai/api/v1/chat/completions",
    headers={
        "Authorization": f"Bearer {openrouter_api_key}",
        "Content-Type": "application/json"
    },
    json={
        "model": "openai/gpt-4o-mini",
        "messages": [
            {
                "role": "system",
                "content": (
                    "Extraia todos os campos exatamente como aparecem "
                    "no documento, incluindo título, autor e contato."
                )
            },
            {
                "role": "user",
                "content": texto_teste
            }
        ],
        "temperature": 0
    },
    timeout=60
)

print("Status HTTP:", resposta.status_code)
print(
    json.dumps(
        resposta.json(),
        indent=4,
        ensure_ascii=False
    )
)
```
