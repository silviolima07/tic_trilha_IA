```python
import os

os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
```

```python
pdf_dir=Path('pdf')
pdf_dir.mkdir(parents=True, exist_ok=True)
#
md_dir=Path('md')

md_dir.mkdir(parents=True, exist_ok=True)

converter = DocumentConverter()

# Localiza todos os PDFs
pdfs = list(pdf_dir.glob("*.pdf"))
```

```python
for pdf in pdfs:
    try:
        print(f"\nConvertendo: {pdf.name}")

        # O convert recebe apenas o PDF
        result = converter.convert(pdf)

        # Converte o resultado para Markdown
        markdown = result.document.export_to_markdown()

        # Define o nome do arquivo de saída
        md_file = md_dir / f"{pdf.stem}.md"

        # Salva o Markdown
        md_file.write_text(markdown, encoding="utf-8")

        print(f"Arquivo salvo: {md_file.resolve()}")

    except Exception as erro:
        print(f"Erro ao converter {pdf.name}: {erro}")

print("\nConversão concluída.")
```
