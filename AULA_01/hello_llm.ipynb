# Introdução à IA - Primeiro Contato com LLMs
Neste notebook, vamos fazer a nossa primeira chamada para a API da OpenAI utilizando as chaves configuradas com segurança.

### 1. Importação de Bibliotecas e Configuração
Primeiro, carregamos as variáveis de ambiente e iniciamos o cliente da OpenAI.

### 2. Chamada à API
Usaremos o modelo configurado no `.env` (ou o `gpt-4o-mini` por padrão) para gerar um texto.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Carrega as variáveis do .env
load_dotenv(override=True)

# Cria o cliente da Groq
client = OpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)

# Lê o modelo do .env
modelo = os.getenv("GROQ_MODEL", "llama-3.3-70b-versatile")

print(f"Modelo: {modelo}")

# Faz a chamada
response = client.chat.completions.create(
    model=modelo,
    messages=[
        {
            "role": "user",
            "content": "Write a haiku about AI."
        }
    ]
)

# Exibe a resposta
print(response.choices[0].message.content)
```
