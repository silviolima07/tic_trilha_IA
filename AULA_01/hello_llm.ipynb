# Introdução à IA - Primeiro Contato com LLMs
Neste notebook, vamos fazer a nossa primeira chamada para a API da OpenAI utilizando as chaves configuradas com segurança.

### 1. Importação de Bibliotecas e Configuração
Primeiro, carregamos as variáveis de ambiente e iniciamos o cliente da OpenAI.

### 2. Chamada à API
Usaremos o modelo configurado no `.env` (ou o `gpt-4o-mini` por padrão) para gerar um texto.

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI

# Carrega uma única vez as variáveis do arquivo .env
load_dotenv(override=True)


# ============================================================
# GROQ
# ============================================================

groq_api_key = os.getenv("GROQ_API_KEY")
groq_modelo = os.getenv(
    "GROQ_MODEL",
    "llama-3.3-70b-versatile"
)

if not groq_api_key:
    raise ValueError("A variável GROQ_API_KEY não foi encontrada no .env")

cliente_groq = OpenAI(
    api_key=groq_api_key,
    base_url="https://api.groq.com/openai/v1"
)

print("="*20)
print("Conectando com a Groq...")
print(f"Modelo: {groq_modelo}")

try:
    resposta_groq = cliente_groq.chat.completions.create(
        model=groq_modelo,
        messages=[
            {
                "role": "user",
                "content": "Write a haiku about AI."
            }
        ]
    )

    print("\nResposta da Groq:")
    print(resposta_groq.choices[0].message.content)

except Exception as erro:
    print("\nErro na Groq:")
    print(erro)

print("\n" + "="*20 + "\n")
# ============================================================
# OPENROUTER
# ============================================================

openrouter_api_key = os.getenv("OPENROUTER_API_KEY")

if not openrouter_api_key:
    raise ValueError(
        "A variável OPENROUTER_API_KEY não foi encontrada no .env"
    )

print("Conectando ao OpenRouter...")

resposta_openrouter = requests.post(
    url="https://openrouter.ai/api/v1/chat/completions",
    headers={
        "Authorization": f"Bearer {openrouter_api_key}",
        "Content-Type": "application/json"
    },
    json={
        "model": "openai/gpt-3.5-turbo",
        "messages": [
            {
                "role": "user",
                "content": (
                    "Olá! Estou pronta para a aula "
                    "de Introdução à IA."
                )
            }
        ]
    },
    timeout=60
)

# Interrompe em erros HTTP, como 401, 404 ou 429
try:
    resposta_openrouter.raise_for_status()
except requests.exceptions.HTTPError:
    print("\nErro HTTP no OpenRouter:")
    print("Status:", resposta_openrouter.status_code)
    print("Resposta:", resposta_openrouter.text)
else:
    resultado_openrouter = resposta_openrouter.json()

    if "choices" in resultado_openrouter:
        print("\nResposta do OpenRouter:")
        print(
            resultado_openrouter["choices"][0]
            ["message"]["content"]
        )

    elif "error" in resultado_openrouter:
        print("\nErro retornado pelo OpenRouter:")
        print(resultado_openrouter["error"])

    else:
        print("\nResposta inesperada do OpenRouter:")
        print(resultado_openrouter)
        
```
